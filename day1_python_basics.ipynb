## 2. Variables and assignment

A **variable** is a named container that holds a value. You create a variable by writing `name = value`.

Let's store some information about a gene:

````python
# Store information about a gene
gene_name = "BRCA1"          # the gene's name (a string)
chromosome = 17               # chromosome number (an integer)
expression_value = 8.43       # expression level in log2 scale (a float)
p_value = 0.0023              # statistical significance (a float)
is_significant = True         # is the result significant? (a boolean)

# Print the values
print(gene_name)
print(chromosome)
print(expression_value)
print(is_significant)
````

Variables can be reassigned at any time. Python simply overwrites the old value:

````python
# Reassign a variable
expression_value = 9.10
print("Updated expression:", expression_value)
````

## 3. Data types

Use the built-in `type()` function to check the type of any variable:

## 4. String operations

In bioinformatics, **DNA and protein sequences are strings**. Python's string tools are therefore extremely useful.

Let's explore:

````python
# A DNA sequence
seq = "atgcgatcgatcgatcgaac"

# Length
print("Length:", len(seq))

# Convert to uppercase (standard for sequences)
seq = seq.upper()
print("Uppercase:", seq)
````

````python
# Slicing — extract a substring
# Python uses 0-based indexing: first character is at position 0
first_codon = seq[0:3]   # characters at positions 0, 1, 2
print("First codon:", first_codon)

last_char = seq[-1]      # -1 means the last character
print("Last base:", last_char)

# Every third character (codons)
print("All codons:")
for i in range(0, len(seq), 3):
    print(" ", seq[i:i+3])
````

````python
# Counting and replacing
print("Count of 'A':", seq.count("A"))
print("Count of 'G':", seq.count("G"))

# Replace T with U to get RNA sequence
rna = seq.replace("T", "U")
print("RNA sequence:", rna)
````

````python
# String concatenation
gene = "BRCA"
number = "1"
full_name = gene + number
print(full_name)

# Check if a motif is present
if "ATG" in seq:
    print("Start codon (ATG) found!")
else:
    print("No start codon found.")
````

## 5. Arithmetic — computing GC content

GC content (the proportion of G and C bases in a DNA sequence) is one of the most basic sequence statistics. Let's compute it step by step.

````python
# Base counts
A = 18
T = 12
G = 20
C = 10

total = A + T + G + C
gc_count = G + C

# GC content as a percentage
gc_content = (gc_count / total) * 100

print("Total bases:", total)
print("GC count:", gc_count)
print("GC content:", gc_content, "%")
````

````python
# Common arithmetic operators in Python
print(10 + 3)   # addition       -> 13
print(10 - 3)   # subtraction    -> 7
print(10 * 3)   # multiplication -> 30
print(10 / 3)   # division       -> 3.333...
print(10 // 3)  # floor division -> 3 (integer result)
print(10 % 3)   # modulo         -> 1 (remainder)
print(2 ** 8)   # exponentiation -> 256
````

---
## 6. Comparisons and booleans

Comparisons return `True` or `False` and are the foundation of all decision-making in code.

````python
# Comparison operators
print(gc_content > 50)   # is GC content above 50%?
print(gc_content == 50)  # exactly 50?
print(gc_content != 50)  # not 50?

# Combining conditions
p = 0.03
fc = 2.5
print(p < 0.05 and fc > 2)   # both conditions must be true
print(p < 0.05 or fc > 2)    # at least one must be true
print(not p < 0.05)           # invert the result
````

---
## 7. f-strings — formatted output

f-strings (formatted string literals) let you embed variables directly into text. They are the modern, readable way to produce output.

````python
gene = "BRCA1"
gc = 45.0
p_val = 0.0023

# Basic f-string
print(f"Gene: {gene}")

# With formatting — :.2f means 'float with 2 decimal places'
print(f"GC content: {gc:.2f}%")

# Scientific notation for small numbers
print(f"p-value: {p_val:.4f}")
print(f"p-value: {p_val:.2e}")   # scientific notation

# Full summary line
print(f"Gene {gene} has a GC content of {gc:.1f}% (p = {p_val:.4f})")
````
