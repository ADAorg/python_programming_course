## 4. String operations

In bioinformatics, **DNA and protein sequences are strings**. Python's string tools are therefore extremely useful.

Let's explore:

````python
# A DNA sequence
seq = "atgcgatcgatcgatcgaac"

# Length
print("Length:", len(seq))

# Convert to uppercase (standard for sequences)
seq = seq.upper()
print("Uppercase:", seq)
````

````python
# Slicing — extract a substring
# Python uses 0-based indexing: first character is at position 0
first_codon = seq[0:3]   # characters at positions 0, 1, 2
print("First codon:", first_codon)

last_char = seq[-1]      # -1 means the last character
print("Last base:", last_char)

# Every third character (codons)
print("All codons:")
for i in range(0, len(seq), 3):
    print(" ", seq[i:i+3])
````

````python
# Counting and replacing
print("Count of 'A':", seq.count("A"))
print("Count of 'G':", seq.count("G"))

# Replace T with U to get RNA sequence
rna = seq.replace("T", "U")
print("RNA sequence:", rna)
````

````python
# String concatenation
gene = "BRCA"
number = "1"
full_name = gene + number
print(full_name)

# Check if a motif is present
if "ATG" in seq:
    print("Start codon (ATG) found!")
else:
    print("No start codon found.")
````

## 5. Arithmetic — computing GC content

GC content (the proportion of G and C bases in a DNA sequence) is one of the most basic sequence statistics. Let's compute it step by step.

````python
# Base counts
A = 18
T = 12
G = 20
C = 10

total = A + T + G + C
gc_count = G + C

# GC content as a percentage
gc_content = (gc_count / total) * 100

print("Total bases:", total)
print("GC count:", gc_count)
print("GC content:", gc_content, "%")
````

````python
# Common arithmetic operators in Python
print(10 + 3)   # addition       -> 13
print(10 - 3)   # subtraction    -> 7
print(10 * 3)   # multiplication -> 30
print(10 / 3)   # division       -> 3.333...
print(10 // 3)  # floor division -> 3 (integer result)
print(10 % 3)   # modulo         -> 1 (remainder)
print(2 ** 8)   # exponentiation -> 256
````

## 6. Comparisons and booleans

Comparisons return `True` or `False` and are the foundation of all decision-making in code.

````python
# Comparison operators
print(gc_content > 50)   # is GC content above 50%?
print(gc_content == 50)  # exactly 50?
print(gc_content != 50)  # not 50?

# Combining conditions
p = 0.03
fc = 2.5
print(p < 0.05 and fc > 2)   # both conditions must be true
print(p < 0.05 or fc > 2)    # at least one must be true
print(not p < 0.05)           # invert the result
````

## 7. f-strings — formatted output

f-strings (formatted string literals) let you embed variables directly into text. They are the modern, readable way to produce output.

````python
gene = "BRCA1"
gc = 45.0
p_val = 0.0023

# Basic f-string
print(f"Gene: {gene}")

# With formatting — :.2f means 'float with 2 decimal places'
print(f"GC content: {gc:.2f}%")

# Scientific notation for small numbers
print(f"p-value: {p_val:.4f}")
print(f"p-value: {p_val:.2e}")   # scientific notation

# Full summary line
print(f"Gene {gene} has a GC content of {gc:.1f}% (p = {p_val:.4f})")
````

## 8b. Comments and code style (PEP 8)

Good code is written for the next person to read it — including your future self. Python's style guide, **PEP 8**, covers the basics:

- `snake_case` for variable and function names (not `camelCase`)
- 4-space indentation
- Spaces around operators: `x = 1 + 2`, not `x=1+2`
- Single-line comments with `#`; multi-line with `"""docstrings"""`
- Write comments that explain **why**, not **what** — the code already says what it does

````python
# PEP 8 — Python's official style guide. Key rules:
# 1. snake_case for variable and function names
# 2. 4-space indentation (never tabs)
# 3. Spaces around operators

# Good style
gene_name = "BRCA1"
gc_threshold = 0.60   # explain WHY: NCBI says >60% indicates a CpG island
expression_value = 8.43

# Bad style (works but hard to read)
# GeneName="BRCA1"
# gC_ThReShOlD=0.60

# Comments: write WHY, not WHAT — the code already says what it does
total_reads = 1_000_000   # 1 million reads — minimum for reliable DEG calls
# Bad comment: # set total_reads to 1000000

# Single-line comment
# Multi-line explanation → use a triple-quoted string (docstring style)
"""
This section normalises read counts by library size.
Reason: raw counts are confounded by sequencing depth.
"""

print(f"Gene: {gene_name}, GC threshold: {gc_threshold:.0%}")
````

## 9. Reading error messages

Python error messages are precise — they tell you the **type** of error and the **exact line** where it happened. Always read the traceback **bottom-up**: the last line is the most important.

The three errors beginners see constantly:

| Error | Cause | Example |
|-------|-------|---------|
| `SyntaxError` | Code isn't valid Python | Missing `:` after `if` |
| `NameError` | Used a variable that doesn't exist | `print(x)` before `x = ...` |
| `TypeError` | Wrong type for an operation | `"age: " + 25` |
| `ValueError` | Right type, wrong value | `int("hello")` |

````python
# Each example below triggers a different error type.
# They are wrapped in try/except so the notebook keeps running.

# 1. NameError — used a variable that was never defined
try:
    print(undefined_variable)
except NameError as e:
    print(f"NameError:  {e}")

# 2. TypeError — wrong type for an operation
try:
    result = "count: " + 42   # can't concatenate str and int
except TypeError as e:
    print(f"TypeError:  {e}")
    print("  Fix: use an f-string → f'count: {42}'")

# 3. ValueError — right type, but the value doesn't make sense
try:
    number = int("abc")       # "abc" cannot be converted to int
except ValueError as e:
    print(f"ValueError: {e}")

print()
print("Key habit: read the traceback BOTTOM-UP — the last line tells you what went wrong.")
print("Errors are information, not failures. Every programmer sees them daily.")
````
